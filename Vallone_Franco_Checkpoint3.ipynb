```python
def recomendado(puntaje):
    if pd.isna(puntaje):
        return 'Sin Calificar'
    if puntaje >= 6.0:
        return 'Recomendado'
    elif puntaje < 6.0:
        return 'No Recomendado'

df['Recomendado_por_la_critica?'] = df['Critic_Score'].apply(recomendado)
```

```python
#Pruebo a ver la nueva columna
df.head()
```

```python
#Creo columna donde si el raiting es E significa que es apto para todo público

df['Apto_para_todo_publico?'] = df['ESRB_Rating'].apply(lambda x: 'Si' if x == 'E' else 'No')
```

```python

#Detecto que si Total Shipped es Nulo, tengo valores Siempre en Global Sales, por lo que reemplazo los valores de total Shipped en Global Sales si los valores son nulos.

df['Global_Sales'] = df['Global_Sales'].fillna(df['Total_Shipped'])
```

##Saneamiento 2

```python
#Borro filas donde Year sea nulo

df.dropna(subset=['Year'])
```

##Filtro

```python
# Filtrar el DataFrame para años anteriores a 1990
display(df[df['Year'] < 1990])
```

##Saneamiento 3

```python
# Elimino las filas donde el año es anterior a 1990
df = df[df['Year'] >= 1990]
```

```python
# Y las primeras filas para confirmar los cambios
display(df.head())
```

```python
# Cuento la cantidad de celdas con 'RP' en la columna 'ESRB_Rating'
count_rp = len(df[df['ESRB_Rating'] == 'RP'])
print(f"Número de filas con ESRB_Rating 'RP': {count_rp}")
```

RP en ESRB_Rating equivale a = Rating pendiente, por lo que, para limpiar y unificar, reemplazo los valores nulos y RP por Sin clasificar

##Saneamiento 4

```python
df['ESRB_Rating'] = df['ESRB_Rating'].fillna('Sin clasificar')
df['ESRB_Rating'] = df['ESRB_Rating'].replace('RP', 'Sin clasificar')
```

```python
display(df.head())
```

```python
cantidad_nulos_esrb = df['ESRB_Rating'].isnull().sum()
print(f"Número de filas con ESRB_Rating nulo: {cantidad_nulos_esrb}")
```

```python
#Creo una columna para filtrar si tengo registros de ventas distribuidos regionalmente
df['Ventas_Distribuidas_Regionalmente?'] = np.where(
    (df['Global_Sales'].notna()) &
    (df['NA_Sales'].isna()) &
    (df['PAL_Sales'].isna()) &
    (df['JP_Sales'].isna()) &
    (df['Other_Sales'].isna()),
    'No', 'Si'
)
display(df.head())
```

##Filtro booleano
